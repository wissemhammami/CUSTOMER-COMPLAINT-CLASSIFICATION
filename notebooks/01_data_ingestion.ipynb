```python
import os
os.chdir(r'C:\Users\Admin\Desktop\CUSTOMER COMPLAINT CLASSIFICATION')
print(os.getcwd())

import pandas as pd

df = pd.read_csv('data/raw/complaints_raw.csv')
df = df[['Consumer complaint narrative', 'Product']]
df = df.dropna(subset=['Consumer complaint narrative'])
df = df.rename(columns={'Consumer complaint narrative': 'text', 'Product': 'label'})

print(df.shape)
print(df['label'].value_counts())
print(df['label'].value_counts(normalize=True).round(3) * 100)

df.to_csv('data/processed/complaints_clean.csv', index=False)
print('Saved to data/processed/complaints_clean.csv')
```
