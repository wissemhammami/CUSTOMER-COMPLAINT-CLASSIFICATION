```python
import os
os.chdir(r'C:\Users\Admin\Desktop\CUSTOMER COMPLAINT CLASSIFICATION')
print(os.getcwd())

import pandas as pd

df = pd.read_csv('data/processed/complaints_clean.csv')

print('Shape:', df.shape)
print(df.dtypes)
print('Missing values:\n', df.isna().sum())
print('Duplicates:', df.duplicated().sum())

df = df.drop_duplicates()
print('After removing duplicates:', df.shape)

df = df[df['text'].str.len() >= 50]
print('After removing short texts:', df.shape)

df = df.reset_index(drop=True)
print(df['label'].value_counts())

df.to_csv('data/processed/complaints_validated.csv', index=False)
print('Saved to data/processed/complaints_validated.csv')
```
